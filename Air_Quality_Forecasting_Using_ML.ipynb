```python
# counting the number of times -200 appears in the data
air_quality_data.isin([-200]).sum(axis=0)
```

**Handling the missing values**

convert all -200 to NaN

Replace all NaN values with mean of that specific coloumn

```python
air_quality_data = air_quality_data.replace(to_replace=-200, value = np.NaN)
```

This shows the actual number of missing values

```python

air_quality_data.tail()
```

```python
# Convert 'Date' and 'Time' columns to datetime objects
air_quality_data['Date'] = pd.to_datetime(air_quality_data['Date'], format='%d/%m/%Y')
air_quality_data['Time'] = pd.to_datetime(air_quality_data['Time'], format='%H.%M.%S').dt.time

# Select only numeric columns for calculating the mean
numeric_columns = air_quality_data.select_dtypes(include=np.number).columns
air_quality_data_mean = air_quality_data[numeric_columns].mean()

# Print the mean values of the numeric columns
print(air_quality_data_mean)
```

```python
# replacing the missing values with mean value
air_quality_data = air_quality_data.fillna(air_quality_data_mean) # Remove the parentheses
```

```python
air_quality_data.tail()
```

**Forecasting with FB Prophet Algorithm**

FB Prophet documentation: https://facebook.github.io/prophet/docs/quick_start.html

```python
# converting the date from DD/MM/YYYY to YYYY-MM-DD

date_info = pd.to_datetime(air_quality_data['Date'])
print(date_info)
```

```python
time_info = air_quality_data['Time']
```

```python
print(time_info)
```

```python
time_info = time_info.apply(lambda x: str(x).replace('.', ':'))
```

```python
print(time_info)
```

```python
print(type(date_info))
print(type(time_info))
```

```python
# combining 2 series to a pandas daatframe
date_time = pd.concat([date_info, time_info], axis=1)
```

```python
# combining date and time

date_time['ds'] = date_time['Date'].astype(str)+ ' ' + date_time['Time'].astype(str)
```

```python
from prophet import Prophet
```

```python
# training the Prophet model

model = Prophet()
model.fit(data)
```

```python
future = model.make_future_dataframe(periods=365, freq='H')
future.tail()
```

```python
forecast = model.predict(future)
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail()
```
